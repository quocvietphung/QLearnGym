```
from qlearning.train import train_agent
from qlearning.evaluate import evaluate_agent
from qlearning.utils import load_qtable

# Huấn luyện agent
q_table, rewards = train_agent(
    env_name='Taxi-v3',
    num_episodes=3000,
    max_steps=100,
    alpha=0.7,
    gamma=0.95,
    epsilon=1.0,
    epsilon_min=0.1,
    epsilon_decay=0.995,
    model_path="../models/qtable.npy",
    plot_path="results/rewards.png"
)

# Đánh giá agent
evaluate_agent(
    q_table_path="../models/qtable.npy",
    env_name='Taxi-v3',
    episodes=100,
    render=True
)
```
